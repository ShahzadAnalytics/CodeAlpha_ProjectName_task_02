```python

# Select numeric columns
corr_cols = ['Sales', 'Quantity', 'Unit price', 'Tax 5%', 'gross income', 'Rating']

# Compute correlation matrix
corr = df[corr_cols].corr()

# Plot heatmap
plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```

## Correlation Matrix – Key Insights

- **Sales and Gross Income** show a perfect positive correlation, as gross income is derived from sales.
- **Sales and Quantity** have a strong positive correlation, indicating that higher quantity sold increases sales.
- **Sales and Unit Price** show a moderate positive relationship.
- **Sales and Tax 5%** are positively correlated since tax depends on sales.
- **Rating** shows little to no correlation with sales-related metrics.

**Overall Insight:**  
Sales are mainly driven by **quantity sold and unit price**, while **customer ratings have minimal impact on revenue**.


```python
# Remove the outliers of the Tax column
Q1 = df['Tax 5%'].quantile(0.25)
Q3 = df['Tax 5%'].quantile(0.75)
# Calculate IQR
IQR = Q3 - Q1
# Define lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df = df[(df['Tax 5%']>=lower_bound)&(df['Tax 5%']<=upper_bound)]
print(df)
```

## Outlier Removal – Tax 5%

The IQR method was applied to the `Tax 5%` column to remove extreme values.
Observations outside the range `Q1 − 1.5 × IQR` and `Q3 + 1.5 × IQR` were treated as outliers.
After removal, 9 records were excluded from the dataset.


```python
# Remove the outliers of the Gross Income column
Q1 = df['gross income'].quantile(0.25)
Q3 = df['gross income'].quantile(0.75)
# Calculate IQR
IQR = Q3 - Q1
# Define lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df_outliers = df[(df['gross income']>=lower_bound)&(df['gross income']<=upper_bound)]
print(df_outliers)
```

## Outlier Analysis – Gross Income

The IQR-based outlier detection identified a small number of extreme values in the `gross income` column. After removing these outliers, the dataset became more consistent, reducing the influence of unusually high or low income values. This helps ensure that subsequent analyses and visualizations better represent typical business performance rather than being skewed by extreme transactions.
